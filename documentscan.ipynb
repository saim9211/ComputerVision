```python
import cv2
import numpy as np

cap = cv2.VideoCapture(0)
cap.set(3, 640)
cap.set(4, 480)

def preprocessing(img):
    img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    img_blur = cv2.GaussianBlur(img_gray, (5, 5), 1)
    img_canny = cv2.Canny(img_blur, 50, 150)

    kernel = np.ones((5, 5), np.uint8)
    img_dilated = cv2.dilate(img_canny, kernel, iterations=1)
    img_eroded = cv2.erode(img_dilated, kernel, iterations=1)

    return img_eroded

def get_contours(img, display_img):
    biggest = np.array([])
    max_area = 0

    contours, hierarchy = cv2.findContours(
        img,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_NONE
    )

    for cnt in contours:
        area = cv2.contourArea(cnt)

        if area > 5000:
            peri = cv2.arcLength(cnt, True)
            approx = cv2.approxPolyDP(cnt, 0.02 * peri, True)

            if len(approx) == 4 and area > max_area:
                biggest = approx
                max_area = area

    return biggest

while True:
    success, frame = cap.read()

    if not success:
        break

    frame = cv2.resize(frame, (640, 480))
    display_img = frame.copy()

    img_threshold = preprocessing(frame)
    biggest = get_contours(img_threshold, display_img)

    if biggest.size != 0:
        cv2.drawContours(display_img, [biggest], -1, (0, 255, 0), 5)

    cv2.imshow("Threshold", img_threshold)
    cv2.imshow("Document Detection", display_img)
    cv2.imshow("Original", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
