# vedio processing 

```python
import cv2
video_path = r"D:\computervision\ComputerVision-1\opencv\VID-20250405-WA0016.mp4"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print(f"Error opening video file: {video_path}")
else:
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            print("End of video stream or error reading frame.")
            break

        cv2.imshow("Video Frame", frame)

        # Press 'q' to exit the video display
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break
```

# vedio web cam processing 

```python
import cv2

cap = cv2.VideoCapture(0)
cap.set(3,640)
cap.set(4,480)
cap.set(10,10000)


while True:
    ret, frame = cap.read()

    if not ret:
        break

    cv2.imshow("Webcam", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

# important function 

```python
import cv2
import numpy as np
img=cv2.imread(r"D:\computervision\ComputerVision-1\opencv\Screenshot 2026-09-30 053603.png")
kernel=np.ones((5,5),np.uint8)
# grey scaling images
img_gry=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
# blur the image
img_blur=cv2.GaussianBlur(img_gry,(13,13),0)
# canny edge detection
img_canny=cv2.Canny(img_blur,100,100)
# dilating the image
img_dilated=cv2.dilate(img_canny,kernel,iterations=1)
# eroding the image
img_eroded=cv2.erode(img_dilated,kernel,iterations=5)
cv2.imshow("Gray Image",img_gry)
cv2.imshow("Blur Image",img_blur)
cv2.imshow("Canny Image",img_canny)
cv2.imshow("Dilated Image",img_dilated)
cv2.imshow("Eroded Image",img_eroded)
cv2.waitKey(0)
```

# shapes and text on images

```python
import cv2
import numpy as np 
img=cv2.imread(r"D:\computervision\ComputerVision-1\opencv\Screenshot 2026-09-30 053603.png")
pt1=np.array([[50,50],[200,50],[200,200],[50,200]])
pt2=np.array([[10,100],[200,50],[220,220],[100,250]])
matirx=cv2.getPerspectiveTransform(pt1.astype(np.float32),pt2.astype(np.float32))
img_output=cv2.warpPerspective(img,matirx,(300,300))
cv2.imshow("Original Image",img_output)
cv2.waitKey(0)
```
