# vedio processing 

```python
import cv2
video_path = r"D:\computervision\ComputerVision-1\opencv\VID-20250405-WA0016.mp4"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print(f"Error opening video file: {video_path}")
else:
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            print("End of video stream or error reading frame.")
            break

        cv2.imshow("Video Frame", frame)

        # Press 'q' to exit the video display
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break
```

# vedio web cam processing 

```python
import cv2

cap = cv2.VideoCapture(0)
cap.set(3,640)
cap.set(4,480)
cap.set(10,10000)


while True:
    ret, frame = cap.read()

    if not ret:
        break

    cv2.imshow("Webcam", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

# important function 

```python
import cv2
import numpy as np
img=cv2.imread(r"D:\computervision\ComputerVision-1\opencv\Screenshot 2026-09-30 053603.png")
kernel=np.ones((5,5),np.uint8)
# grey scaling images
img_gry=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
# blur the image
img_blur=cv2.GaussianBlur(img_gry,(13,13),0)
# canny edge detection
img_canny=cv2.Canny(img_blur,100,100)
# dilating the image
img_dilated=cv2.dilate(img_canny,kernel,iterations=1)
# eroding the image
img_eroded=cv2.erode(img_dilated,kernel,iterations=5)
cv2.imshow("Gray Image",img_gry)
cv2.imshow("Blur Image",img_blur)
cv2.imshow("Canny Image",img_canny)
cv2.imshow("Dilated Image",img_dilated)
cv2.imshow("Eroded Image",img_eroded)
cv2.waitKey(0)
```

```python
import cv2
import numpy as np 
img=cv2.imread(r"D:\computervision\ComputerVision-1\opencv\Screenshot 2026-09-30 053603.png")
pt1=np.array([[50,50],[200,50],[200,200],[50,200]])
pt2=np.array([[10,100],[200,50],[220,220],[100,250]])
matirx=cv2.getPerspectiveTransform(pt1.astype(np.float32),pt2.astype(np.float32))
img_output=cv2.warpPerspective(img,matirx,(300,300))
cv2.imshow("Original Image",img_output)
cv2.waitKey(0)
```

# color detection

```python
import cv2
import numpy as np

def empty(a):
    pass
cv2.namedWindow("TrackBars")
cv2.resizeWindow("TrackBars",640,240)
cv2.createTrackbar("Hue Min","TrackBars",0,179,empty)
cv2.createTrackbar("Hue Max","TrackBars",179,179,empty)
cv2.createTrackbar("Sat Min","TrackBars",0,255,empty)
cv2.createTrackbar("Sat Max","TrackBars",255,255,empty)
cv2.createTrackbar("Val Min","TrackBars",0,255,empty)
cv2.createTrackbar("Val Max","TrackBars",255,255,empty)

while True:
    path=r"D:\computervision\ComputerVision-1\opencv\Screenshot 2026-09-30 053603.png"
    img=cv2.imread(path)
    img_hsv=cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
    h_min=cv2.getTrackbarPos("Hue Min","TrackBars")
    h_max=cv2.getTrackbarPos("Hue Max","TrackBars")
    s_min=cv2.getTrackbarPos("Sat Min","TrackBars")
    s_max=cv2.getTrackbarPos("Sat Max","TrackBars")
    v_min=cv2.getTrackbarPos("Val Min","TrackBars")
    v_max=cv2.getTrackbarPos("Val Max","TrackBars")

    lower=np.array([h_min,s_min,v_min])
    higher=np.array([h_max,s_max,v_max])

    mask=cv2.inRange(img_hsv,lower,higher)
    img_out=cv2.bitwise_and(img,img,mask=mask)
    cv2.imshow("Original Image",img)
    cv2.imshow("HSV Image",img_hsv)
    cv2.imshow("Output Image",img_out)
    if cv2.waitKey(1) & 0xFF==ord("q"):
        break
cv2.destroyAllWindows()
```
